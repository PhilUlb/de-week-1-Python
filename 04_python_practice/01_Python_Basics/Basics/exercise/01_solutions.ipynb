#### 1. Type explorer

```python
def num_type(x):
    return type(x).__name__

assert num_type(3) == "int"
assert num_type(2.0) == "float"
assert num_type(True) == "bool"
```

#### 2. Safe division

```python
def safe_div(a, b, default=None):
    if b==0:
        return default
    return a/b
    
assert safe_div(6, 3) == 2.0
assert safe_div(1, 0, default=float('inf')) == float('inf')
```

#### 3. Integer vs float division

```python
def divisions(a, b):
    int_div = a//b
    reg_div = a/b
    mod_div = a%b
    return (int_div, reg_div, mod_div)
    
assert divisions(7, 3) == (2, 7/3, 1)
assert divisions(-7, 3)[0] == -3   # floor division
```

#### 4. Rounding modes

```python
from math import floor

def bankers_round(x, ndigits=0):
    return round(x, ndigits)

def classic_round(x, ndigits=0):
    factor = 10 ** ndigits
    return floor(x * factor + 0.5) / factor

assert bankers_round(2.5) == 2
assert classic_round(2.5) == 3
```

#### 5. Parse numbers

```python
def parse_int(s, base=10, default=None):
    s = s.strip()
    try:
        return int(s, base)
    except ValueError as e:
            return default

assert parse_int(" 1_024 ") == 1024
assert parse_int("FF", base=16) == 255
assert parse_int("oops", default=-1) == -1
```

#### 6. Boolean arithmetic

```python
def true_ratio(seq):
    return sum([bool(x) for x in seq])/len(seq)

assert abs(true_ratio([True, False, 1, 0, "", "x"]) - 0.5) < 1e-9
```

#### 7. Min/Max & aggregates

```python
def stats(nums):
    result = {}
    result["count"] = len(nums)
    result["total"] = sum(nums)
    result["mean"] = result["total"] / result["count"]
    result["minimum"] = min(nums)
    result["maximum"] = max(nums)
    return result

assert stats([1,2,3])["mean"] == 2
```

#### 8. Bucket by step

```python
def bucket(x, step):
    return x - x % step
assert bucket(37, 10) == 30
assert bucket(-3, 5) == -5
```

#### 9. Time math (seconds → h\:m\:s)

```python
def quotient_and_remainder(x, divisor):
    quotient = x // divisor
    remainder = x % divisor
    return (quotient, remainder)

def to_hms(seconds):
    hours, remainder = quotient_and_remainder(seconds, 3600)
    minutes, seconds = quotient_and_remainder(remainder, 60)
    return f"{hours:02d}:{minutes:02d}:{seconds:02d}"
    
assert to_hms(3661) == "01:01:01"
```

#### 10. Running totals with precise rounding

```python
def running_totals(amounts, ndigits=2):
    for i,x in enumerate(amounts):
        tmp = round(x, ndigits)
        if i==0:
            result = [tmp]
        else:
            result.append(result[i-1] + tmp)
    return result

assert running_totals([10.005, 0.005], 2) == [10.01, 10.02]
```
