### Warm-ups

1. **FizzBuzz (classic)**

   * `fizzbuzz(n)` → `"Fizz"`, `"Buzz"`, `"FizzBuzz"`, or `str(n)`. If n is multiple of 3: `Fizz`; multiple of 5: `Buzz`; multiple of 3 and 5: `FizzBuzz`; else: n itself

````python
def fizzbuzz(n):
    if not n % 15: return "FizzBuzz" # 15 = lowest common denominator of 3 and 5
    if not n % 3: return "Fizz"
    if not n % 5: return "Buzz"
    return str(n)
assert fizzbuzz(15) == "FizzBuzz"
````

2. **Categorize temperature**

   * `temp_label(c)` → `freezing` (<0), `cold` (<10), `mild` (<20), `warm` (<30), `hot` (else).

````python
def temp_label(c):
    if c < 0: return "freezing"
    if c < 10: return "cold"
    if c < 20: return "mild"
    if c < 30: return "warm"
    return "hot"

assert temp_label(-20) == "freezing"
assert temp_label(5) == "cold"
assert temp_label(12) == "mild"
assert temp_label(25) == "warm"
assert temp_label(35) == "hot"
````

3. **Truthy length**

   * `choose_label(s)` → `"empty"` if string is empty/whitespace; `"short"` (≤3), `"long"` (>3).

````python
def choose_label(s):
    if not s.split(): # default splitting returns empty list if there's only whitespace
        return "empty"
    if len(s) < 4: return "short"
    return "long"

assert choose_label("") == "empty"
assert choose_label("   ") == "empty"
assert choose_label("a") == "short"
assert choose_label("abc") == "short"
assert choose_label("abcd") == "long"
````

4. **Grade boundaries**

   * `grade(score)` with 0–100 validation; raise `ValueError` otherwise. Return `A/B/C/D/F`.

````python
def grade(score):
    try:
        score = int(score)
    except ValueError as e:
        raise ValueError
    if score < 0 or score > 100:
        raise ValueError
    
    if score < 60: return "F"
    if score < 70: return "D"
    if score < 80: return "C"
    if score < 90: return "B"
    return "A"

# valid range
assert grade(95) == "A"
assert grade(85) == "B"
assert grade(75) == "C"
assert grade(65) == "D"
assert grade(55) == "F"

# edge boundaries
assert grade(100) == "A"
assert grade(90) == "A"
assert grade(89) == "B"
assert grade(80) == "B"
assert grade(79) == "C"
assert grade(70) == "C"
assert grade(69) == "D"
assert grade(60) == "D"
assert grade(59) == "F"
assert grade(0) == "F"
````

5. **Password strength (simple)**

   * `password_strength(s)` returns `"weak"`, `"medium"`, `"strong"` based on length and presence of digit/special.

    * "weak": less than 6 characters, or only letters or only digits

    * "medium": at least 6 characters, with both letters and digits

    * "strong": at least 8 characters, with letters, digits, and special characters

````python
import string

def password_strength(s):
    letters = string.ascii_letters
    digits = "1234567890"
    special = string.punctuation

    has_only_letters = all(e in letters for e in s)
    has_only_digits = all(e in digits for e in s)
    has_alphanum = not has_only_digits and not has_only_letters
    has_special_chars = any(e in special for e in s)

    if len(s) > 7 and has_special_chars and has_alphanum:
        return "strong"
    if len(s) < 6 or (has_only_digits or has_only_letters):
        return "weak"
    return "medium"

# weak: < 6 chars
assert password_strength("abc") == "weak"
assert password_strength("12345") == "weak"

# weak: only letters (even if long)
assert password_strength("abcdef") == "weak"
assert password_strength("abcdefghijk") == "weak"

# weak: only digits (even if long)
assert password_strength("123456") == "weak"
assert password_strength("987654321") == "weak"

# medium: ≥6 chars, letters + digits, no special required
assert password_strength("abc123") == "medium"
assert password_strength("A1b2C3") == "medium"
assert password_strength("hello1") == "medium"

# strong: ≥8 chars, letters + digits + special
assert password_strength("Abc123!!") == "strong"
assert password_strength("pass99$X") == "strong"
assert password_strength("Test1234!") == "strong"
````
