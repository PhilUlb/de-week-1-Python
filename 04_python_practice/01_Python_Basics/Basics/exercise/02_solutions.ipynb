### Warm-ups

1. **Middle char(s)**

   * `middle(s)` → If odd length return the middle char, else return the two middle chars.

```python
def middle(s):
    result = s[len(s)//2]
    if len(s) % 2 == 0:
        result = s[len(s)//2 - 1] + result
    return result

assert middle("abc") == "b"
assert middle("abba") == "bb"
```

2. **Title-case safely**

   * `safe_title(s)` → title case but do not lowercase words fully in ALL CAPS acronyms (e.g., `"learn SQL now"` → `"Learn SQL Now"`; `"use GPU"` → `"Use GPU"`).

```python
def safe_title(s):
    return " ".join([e if e.isupper() else e.capitalize() for e in s.split(" ")])

assert safe_title("learn SQL now") == "Learn SQL Now"
assert safe_title("use GPU") == "Use GPU"
```

3. **Reverse words**

   * `reverse_words(s)` → reverse word order but keep internal characters.

```python
def reverse_words(s):
    return " ".join([e for e in reversed(s.split(" "))])

assert reverse_words("one two three") == "three two one"
```

### Core

4. **Normalize spaces**

   * `squash_spaces(s)` → collapse multiple spaces/tabs to a single space; strip ends.

```python
def squash_spaces(s):
    return " ".join(s.split())
assert squash_spaces("  a\t b   c ") == "a b c"
```

5. **CSV line → list (no quotes)**

   * `split_csv(s)` → split by commas, trim whitespace around fields; ignore empty trailing field if line ends with comma.

```python
def split_csv(s):
    tmp = [e.strip() for e in s.split(",")]
    return [e for e in tmp if e]

assert split_csv(" a, b ,c,") == ["a","b","c"]
```

6. **Mask secrets**

   * `mask_email(s)` → keep first char of user and domain, mask the rest with `*`, keep TLD.

     * `"alice@example.com"` → `"a****@e******.com"`

```python
def mask_email(s):
    def mask(s):
        return s[0] + "*" * (len(s) -1)
    
    rest, tld = s.rsplit(".",1)
    usr, dmn = rest.split("@")
    usr_msk = mask(usr)
    dmn_msk = mask(dmn)

    return f"{usr_msk}@{dmn_msk}.{tld}"

assert mask_email("alice@example.com") == "a****@e******.com"
```

7. **Find all indexes**

   * `find_all(s, sub)` → list of start indices where `sub` occurs (including overlaps).

```python
def find_all(s, sub):
    result = []
    start_idx = 0
    while True:
        i = s.find(sub, start_idx)
        if i == -1:
            return result
        result.append(i)
        start_idx = i +1

assert find_all("aaaa", "aa") == [0,1,2]
```

8. **Anagram check**

* `is_anagram(a, b)` ignoring spaces, case, punctuation.[Anagram](https://en.wikipedia.org/wiki/Anagram)

```python
import string

def is_anagram(a, b):
    del_chars = string.punctuation + " " # to remove any punctuation and whitespace
    a_clean = "".join(e for e in sorted(a.lower()) if e not in del_chars)
    b_clean = "".join(e for e in sorted(b.lower()) if e not in del_chars)
    return a_clean == b_clean

assert is_anagram("Listen", "Silent")
assert is_anagram("Tom Marvolo Riddle", "I am Lord Voldemort!")
```

9. **Format table row**

   * `fmt_row(values, widths)` → left-align strings to fixed widths, joined by `" | "`.

```python
def fmt_row(values, widths):
    out = []
    for v, w in zip(values, widths):
        out.append(v.ljust(w, " "))
    return " | ".join(out)

assert fmt_row(["a","bb"], [3,4]) == "a   | bb  "
```

### Challenge

10. **Slugify**

    * `slugify(title)` → lowercase, trim, replace runs of non-alnum with single `-`, remove leading/trailing `-`.

```python
def slugify(title):
    alphanum = string.ascii_lowercase + "1234567890"
    s = "".join(e if e in alphanum else "-" for e in title.lower())

    # Recursively replace "--" with "-" until only "-" remain (slow, would use regex in prod)
    while True:
        i = s.find("--")
        if i == -1:
            break
        s = s.replace("--", "-")
    
    if s[0] == "-": s = s[1:]
    if s[-1] == "-": s = s[:-1]

    return s

assert slugify("Hello,  World!!") == "hello-world"
```
